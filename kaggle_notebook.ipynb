```python
import pandas as pd
df=pd.read_csv("/kaggle/input/datasets/akshat3422/my_ata/clinical_texts.csv")
```

```python
import torch

print("CUDA available:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "No GPU")
```

```python
BATCH_SIZE = 10_000
OUTPUT_FILE="/kaggle/working/embedding"
MODEL_BATCH_SIZE = 128
from sentence_transformers import SentenceTransformer
import time 



device = "cuda" if torch.cuda.is_available() else "cpu"

model = SentenceTransformer(
    "NeuML/bioclinical-modernbert-base-embeddings",
    device=device
)

print("Using device:", device)

print("Embedding dimension:",
      model.get_sentence_embedding_dimension())

total_start = time.perf_counter()

first_batch = True

for start in range(0, len(df), BATCH_SIZE):

    end = min(start + BATCH_SIZE, len(df))

    batch = df.iloc[start:end]

    print(f"\nProcessing {start:,} → {end:,}")

    batch_start = time.perf_counter()

    embeddings = model.encode(
        batch["clinical_text"].fillna("").tolist(),
        batch_size=MODEL_BATCH_SIZE,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    result = pd.DataFrame({
        "id": batch["id"].values,
        "embedding": [
            "[" + ",".join(map(str, emb)) + "]"
            for emb in embeddings
        ]
    })

    result.to_csv(
        OUTPUT_FILE,
        mode="w" if first_batch else "a",
        header=first_batch,
        index=False
    )

    first_batch = False

    elapsed = time.perf_counter() - batch_start

    print(f"Batch time: {elapsed/60:.2f} min")
    print(f"Speed: {(end-start)/elapsed:.2f} rows/sec")


total_time = time.perf_counter() - total_start

print("\n======================")
print("COMPLETE")
print("======================")
print(f"Total rows: {len(df):,}")
print(f"Total time: {total_time/60:.2f} minutes")
print(f"Total time: {total_time/3600:.2f} hours")
print(f"Saved to: {OUTPUT_FILE}")
```
